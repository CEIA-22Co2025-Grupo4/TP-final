## Calculo de distancias

```python
# Ver tipos de columnas lat/long
print("dtypes en crimes (lat/lon):", df_crimes[['Latitude','Longitude']].dtypes)
print("dtypes en estaciones (lat/lon):", df_police_stations[['LATITUDE','LONGITUDE']].dtypes)

# --- 2. Limpiar / preparar latitudes y longitudes ---

# En crímenes: eliminar filas sin latitud o longitud válidas
df_crimes_dist = df_crimes.dropna(subset=["Latitude","Longitude"])
df_crimes_dist["Latitude"] = pd.to_numeric(df_crimes_dist["Latitude"], errors="coerce")
df_crimes_dist["Longitude"] = pd.to_numeric(df_crimes_dist["Longitude"], errors="coerce")
df_crimes_dist = df_crimes_dist.dropna(subset=["Latitude","Longitude"])

# En estaciones similar
df_police_stations_dist = df_police_stations.dropna(subset=["LATITUDE","LONGITUDE"])
df_police_stations_dist["Latitude"] = pd.to_numeric(df_police_stations_dist["LATITUDE"], errors="coerce")
df_police_stations_dist["Longitude"] = pd.to_numeric(df_police_stations_dist["LONGITUDE"], errors="coerce")
df_police_stations_dist = df_police_stations_dist.dropna(subset=["LATITUDE","LONGITUDE"])

# --- 3. Convertir a GeoDataFrames con geometría Point ---

gdf_crimes = gpd.GeoDataFrame(
    df_crimes_dist,
    geometry = df_crimes_dist.apply(lambda r: Point(r["Longitude"], r["Latitude"]), axis=1),
    crs = "EPSG:4326"   # lat/lon estándar
)

gdf_police_stations_dist = gpd.GeoDataFrame(
    df_police_stations_dist,
    geometry = df_police_stations_dist.apply(lambda r: Point(r["Longitude"], r["Latitude"]), axis=1),
    crs = "EPSG:4326"
)

# --- 4. Elegir un CRS proyectado en metros apropiado para la zona de Chicago ---

gdf_crimes_proj = gdf_crimes.to_crs(epsg=26971)
gdf_police_stations_proj = gdf_police_stations_dist.to_crs(epsg=26971)

# --- 5. Calcular la estación más cercana y la distancia ---

# Usamos sjoin_nearest de GeoPandas
gdf_joined = gpd.sjoin_nearest(
    gdf_crimes_proj,
    gdf_police_stations_proj,
    how="left",
    distance_col="dist_to_station"
)

# Ahora en gdf_joined tienes columnas de estaciones (por ejemplo nombre, id) y la distancia
print("Columnas tras join:", gdf_joined.columns)
```

## Creación de mapas

```python
# ——————————————————————————————
# 1. Cargar / preparar datos
# ——————————————————————————————

# Detecta columnas datetime nativas
datetime_cols = [c for c in df_com_areas.columns if pd.api.types.is_datetime64_any_dtype(df_com_areas[c])]
# Detecta columnas object que puedan contener pd.Timestamp dentro
obj_ts_cols = [c for c in df_com_areas.columns
               if df_com_areas[c].dtype == "object" and df_com_areas[c].apply(lambda x: isinstance(x, pd.Timestamp)).any()]

cols_to_convert = list(dict.fromkeys(datetime_cols + obj_ts_cols))  # remover duplicados manteniendo orden
print("Columnas a convertir a str:", cols_to_convert)

for c in cols_to_convert:
    df_com_areas[c] = df_com_areas[c].astype(str)  # convierte NaT a 'NaT' y Timestamp a ISO string

# --- 2b) Eliminar columnas de metadatos/voluminosas ---
for drop_col in (":id", ":version", ":created_at", ":updated_at", "shape_area", "shape_len"):
    if drop_col in df_com_areas.columns:
        df_com_areas = df_com_areas.drop(columns=[drop_col])
        print(f"Eliminada columna: {drop_col}")

# --- 3) Asegurar CRS WGS84 (lat/lon) para Folium (EPSG:4326) ---
# Si com_areas.crs es None y sabés que las coordenadas son lon/lat, descomentá la línea set_crs
if df_com_areas.crs is None:
    # Atención: usar set_crs sólo si realmente las geometrías están en lon/lat.
    com_areas = df_com_areas.set_crs(epsg=4326)
else:
    com_areas = df_com_areas.to_crs(epsg=4326)

geojson_str = com_areas.to_json()

# Separar crímenes con arresto / sin arresto
# Detectar nombre de columna de arresto (dependiendo del CSV)
gdf_result = gdf_joined.copy()
if "Arrest" in gdf_joined.columns:
    gdf_result["Arrest_flag"] = gdf_joined["Arrest"].apply(lambda v: True if str(v).lower() in ("true","t","y","yes","1") else False)
else:
    gdf_result["Arrest_flag"] = False

gdf_arrest = gdf_result[gdf_result["Arrest_flag"] == True]
gdf_noarrest = gdf_result[gdf_result["Arrest_flag"] == False]

# ——————————————————————————————
# 3. Crear el mapa con folium y añadir capas
# ——————————————————————————————

# Mapa base centrado en Chicago
m = folium.Map(location=[41.85, -87.65], zoom_start=11, tiles="CartoDB positron")

# Capa 1: Community Areas con número
def style_community(feature):
    return {
        "fillColor": "none",
        "color": "black",
        "weight": 1,
        "opacity": 0.7
    }

folium.GeoJson(
    geojson_str,
    name="Community Areas",
    style_function=style_community,
    tooltip=folium.GeoJsonTooltip(fields=["area_numbe", "community"],
                                  aliases=["Area #", "Community Name"])
).add_to(m)

# Además, agregar un texto (etiqueta) con el número del área en el centroide
for _, row in df_com_areas.iterrows():
    centroid = row.geometry.centroid
    folium.map.Marker(
        [centroid.y, centroid.x],
        icon=DivIcon(
            icon_size=(150,36),
            icon_anchor=(0,0),
            html=f'<div style="font-size:8pt; color:gray">{row["area_numbe"]}</div>'
        )
    ).add_to(m)

# Capa 2: Estaciones de policía
cluster_police_stations = MarkerCluster(name="Estaciones de Policia").add_to(m)
for _, row in gdf_police_stations_dist.iterrows():
    folium.CircleMarker(
        location=[row.geometry.y, row.geometry.x],
        radius=4,
        color="blue",
        fill=True,
        fill_color="blue",
        fill_opacity=0.7,
        popup=f"Station: {row.get('station_name', row.get('name', 'N/A'))}"
    ).add_to(cluster_police_stations)

# Capa 3: Crímenes con arresto (puntos verdes)
cluster_arrest = MarkerCluster(name="Crímenes con arresto").add_to(m)
for _, r in gdf_arrest.iterrows():
    folium.CircleMarker(
        location=[r.geometry.y, r.geometry.x],
        radius=3,
        color="green",
        fill=True,
        fill_opacity=0.6,
        popup=f"Crime ID: {r.get('ID','')}<br>Distance to station: {r.get('dist_to_station',0):.1f} m"
    ).add_to(cluster_arrest)

# Capa 4: Crímenes sin arresto (puntos rojos)
cluster_no = MarkerCluster(name="Crímenes sin arresto").add_to(m)
for _, r in gdf_noarrest.iterrows():
    folium.CircleMarker(
        location=[r.geometry.y, r.geometry.x],
        radius=3,
        color="red",
        fill=True,
        fill_opacity=0.4,
        popup=f"Crime ID: {r.get('ID','')}<br>Distance to station: {r.get('dist_to_station',0):.1f} m"
    ).add_to(cluster_no)

# Control de capas
folium.LayerControl(collapsed=False).add_to(m)

# Guardar como html
m.save("mapa_completo_chicago.html")
print("Mapa guardado: mapa_completo_chicago.html")
```

## Conversion de columnas

```python
gdf_joined.columns
```

```python
# --- 6. Seleccionar/renombrar columnas de interés ---

cols_to_keep = [
    "Date", "IUCR", "Primary Type",
    "Location Description",
    "Arrest", "Domestic", "Beat",
    "District", "Ward", "Community Area",
    "FBI Code", "X Coordinate",	"Y Coordinate",	"Latitude_left",	"Longitude_right"	,
    "DISTRICT", "DISTRICT NAME", "dist_to_station"
]
gdf_selected = gdf_joined[cols_to_keep].copy()

# --- 7. Ver algunas filas resultado ---
print("Resultado — muestras con estación más cercana y distancia:")
display(gdf_selected.head())
```

```python
# Renombramos columnas para mayor claridad
gdf_selected["Latitude"] = gdf_selected["Latitude_left"]
gdf_selected["Longitude"] = gdf_selected["Longitude_right"]
gdf_selected["Crime District"] = gdf_selected["District"]
gdf_selected["Nearest Police Station District"] = gdf_selected["DISTRICT"]
gdf_selected["Nearest Police Station District Name"] = gdf_selected["DISTRICT NAME"]
gdf_selected["Distance Crime To Police Station"] = gdf_selected["dist_to_station"]
gdf_selected = gdf_selected.drop(["District", "DISTRICT", "DISTRICT NAME", "dist_to_station", "Longitude_right", "Latitude_left"], axis=1)
gdf_selected.head()
```

```python
# Convertimos columna date a formato datetime
gdf_selected["Date"] = pd.to_datetime(gdf_selected["Date"])
print(f"Columnas del dataframe:\n{gdf_selected.dtypes}\n")
```

```python
# Método auxiliar para determinar la temporada del año
def get_year_season(month_number: int) -> str:
  if month_number in [12, 1, 2]:
      return "Winter"
  elif month_number in [3, 4, 5]:
      return "Spring"
  elif month_number in [6, 7, 8]:
      return "Summer"
  elif month_number in [9, 10, 11]:
      return "Autumn"
  else:
      return "Invalid Month Number"
```

```python
# Método auxiliar para determinar la etapa del día
def get_day_stage_x4(hour_number: int) -> str:
  if hour_number in [0, 1, 2, 3, 4, 5]:
      return "Early Morning"
  elif hour_number in [6, 7, 8, 9, 10, 11]:
      return "Morning"
  elif hour_number in [12, 13, 14, 15, 16, 17]:
      return "Afternoon"
  elif hour_number in [18, 19, 20, 21, 22, 23]:
      return "Night"
  else:
      return "Invalid Hour Number"
```

```python
# Cáculamos los valores de las nuevas columnas
crime_day_array = np.empty(0, dtype=str)
crime_hour_array = np.empty(0, dtype=str)
crime_season_array = np.empty(0, dtype=str)
for _, row in gdf_selected.iterrows():
  crime_day_array = np.append(crime_day_array, row.get("Date").day_name())
  hour_number = row.get("Date").hour
  crime_hour_array= np.append(crime_hour_array, get_day_stage_x4(hour_number))
  month_number = row.get("Date").month
  crime_season_array = np.append(crime_season_array, get_year_season(month_number))
```

```python
# Agregamos las nuevas columnas
gdf_selected["Season"] = crime_season_array
gdf_selected["Day"] = crime_day_array
gdf_selected["Day Time"] = crime_hour_array
```

```python
# Eliminamos las columnas que ya no serán de utilidad
gdf_selected = gdf_selected.drop("Date", axis=1)
gdf_selected.head(10)
```
